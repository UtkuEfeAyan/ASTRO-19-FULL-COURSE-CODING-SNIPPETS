```python
import math
import numpy as np
from astropy.table import Table

def sin_x(x):
    #returns the sine of x radi
    return math.sin(x)
```

```python
def cos_x(x):
    #ceturns the cosine of x radi
    return math.cos(x)
```

```python
x_values = np.linspace(0, 2 * np.pi, 1000)

sin_values = [sin_x(x) for x in x_values]
cos_values = [cos_x(x) for x in x_values]
```

```python
table = Table(
    [x_values, sin_values, cos_values],
    names=('x', 'sin(x)', 'cos(x)')
)

print("Full tabulation:")
print(table)
```
