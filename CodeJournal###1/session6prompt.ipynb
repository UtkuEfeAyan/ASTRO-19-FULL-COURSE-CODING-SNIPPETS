```python
import math
def sin_x(x):
    #returns the sine of x radi
    return math.sin(x)
```

```python
def cos_x(x):
    #ceturns the cosine of x radi
    return math.cos(x)
```

```python
import numpy as np
# create x values from 0 to 2 with 1000 entries
x_values = np.linspace(0, 2 * np.pi, 1000)

# compute sin(x) and cos(x) using the defined functions (
sin_values = [sin_x(x) for x in x_values]
cos_values = [cos_x(x) for x in x_values]
```

```python
import pandas as pd

# create a dataframe to tabulate the results
df = pd.DataFrame({
    'x': x_values,
    'sin(x)': sin_values,
    'cos(x)': cos_values
})

print("Full tabulation:")
print(df)
```
