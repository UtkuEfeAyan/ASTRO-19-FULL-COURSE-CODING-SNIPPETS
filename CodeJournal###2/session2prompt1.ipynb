```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
```

```python
def calc_exponential(x, a=5.0, b=1.2):
    return a * np.exp(-b * x)
    
# Set y values
y = calc_exponential(x)
```

```python
#figure and axis
f, ax = plt.subplots(1, 1, figsize=(6, 4))

#lot x vs. y
ax.plot(x, y, color='blue', linewidth=2.0)

# Labels. of axes
ax.set_xlabel('x', fontsize=14)
ax.set_ylabel('y = a*exp(-bx)', fontsize=14)

# Save the figure as a PDF
plt.savefig('exponential_decay_plot.pdf', bbox_inches='tight')
```
