```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# generate  x data and clean base y data
npoints = 50
x = np.linspace(0, 10, npoints)
y_clean = 2.0 * x + 1.0 
```

```python
# non-constant sigma array, decresases uncertainty 
sigma = np.linspace(2.0, 0.2, npoints)
```

```python
# add scatter to data
scatter = np.random.normal(loc=0.0, scale=sigma, size=npoints)
y_new = y_clean + scatter
```

```python
plt.errorbar(x, y_new, yerr=sigma, fmt='^', color='purple', ecolor='gray', label='Data with variable scatter')

# axis labels
plt.xlabel("X")
plt.ylabel("Y (with decreasing scatter)")
plt.legend()

# save the figure as a PDF file also ensure the labels aren't cut off
plt.savefig("variable_scatter_plot.pdf", bbox_inches='tight')

# Show the plot on the screen
plt.show()
```
