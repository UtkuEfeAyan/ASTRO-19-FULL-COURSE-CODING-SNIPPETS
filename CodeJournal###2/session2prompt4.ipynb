```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Poisson distribution my choice cause sounded fun and different from the normal distribution used in the lecture examples
np.random.seed(42) # seed for reproducibility
random_numbers = np.random.poisson(lam=20.0, size=1000)
```

```python
f, ax = plt.subplots(1, 1, figsize=(8, 5)) 
# put the random numbers into 100 bins and plot
ax.hist(random_numbers, bins=100, facecolor='blue', edgecolor='black', alpha=0.7) 

# Label the axes
ax.set_xlabel('Value (Number of Events)', fontsize=14) 
ax.set_ylabel('Frequency', fontsize=14)
ax.set_title('Poisson Distribution (1000 samples, 100 bins)', fontsize=16)

#Save the figure as a PDF
plt.savefig('poisson_histogram.pdf', bbox_inches='tight', dpi=400)

plt.show()
```
